```python
import pandas as pd
```

```python
with open('../Data/10_computable_moments.txt', 'r') as file:
    lines = file.read().strip().split('\n')

lines = lines[1:]
```

```python
parsed = []
index = 0
for line in lines:
    parts = line.strip().split()
    painting_type = parts[0]
    num_tags = int(parts[1])
    tags = set(parts[2:]) 
    parsed.append({
        "index" : index,
        "Type": painting_type,
        "Num_Tags": num_tags,
        "Tags": tags
    })
    index += 1
df = pd.DataFrame(parsed)
df = df.sort_values(by='Type', ascending=False)


df_p = df[df['Type'] == 'P'].copy()
df_l = df[df['Type'] == 'L'].copy()
merged = []
for i in range(0, len(df_p), 2):
    if i + 1 < len(df_p):
        idx1, idx2 = df_p.iloc[i]['index'], df_p.iloc[i+1]['index']
        tags1, tags2 = df_p.iloc[i]['Tags'], df_p.iloc[i+1]['Tags']
        merged.append({
            "index": f"{idx1} {idx2}",
            "Type" : "P",
            "Tags": tags1.union(tags2)
        })

merged_df = pd.DataFrame(merged)

combined_df = pd.concat([merged_df, df_l], ignore_index=True)

output = combined_df['index']  
len_output = len(output)

with open("../output/10_computable_moments.txt", "w") as f:
    f.write(str(len_output) + "\n")
    for line in output.values:
        f.write(str(line)+'\n')


output.head(20)
```
