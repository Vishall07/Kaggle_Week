```python
import pandas as pd
import itertools
```

```python
input_paths = [
    "../Data/0_example.txt",
    "../Data/1_binary_landscapes.txt",
    "../Data/10_computable_moments.txt",
    "../Data/11_randomizing_paintings.txt",
    "../Data/110_oily_portraits.txt"
    ]

output_paths = [
        "../output/0_example.txt",
        "../output/1_binary_landscapes.txt",
        "../output/10_computable_moments.txt",
        "../output/11_randomizing_paintings.txt",
        "../output/110_oily_portraits.txt"
    ]
```

```python
file_number = 2

input_path = input_paths[file_number]
output_path = output_paths[file_number]

with open(input_path, 'r') as file:
    lines = file.read().strip().split('\n')
lines = lines[1:]
```

```python

parsed = []
index = 0
for line in lines:
    parts = line.strip().split()
    painting_type = parts[0]
    num_tags = int(parts[1])
    tags = set(parts[2:]) 
    parsed.append({
        "index" : index,
        "Type": painting_type,
        "Num_Tags": num_tags,
        "Tags": tags
    })
    index += 1
df = pd.DataFrame(parsed)
print(df)
```

```python

df_p = df[df['Type'] == 'P'].copy()
df_l = df[df['Type'] == 'L'].copy()
merged = []
for i in range(0, len(df_p), 2):
    if i + 1 < len(df_p):
        idx1, idx2 = df_p.iloc[i]['index'], df_p.iloc[i+1]['index']
        tags1, tags2 = df_p.iloc[i]['Tags'], df_p.iloc[i+1]['Tags']
        tags3 = tags1.union(tags2)
        merged.append({
                "index": f"{idx1} {idx2}",
                "Type" : "P",
                "Tags": tags3,
                "Num_Tags" : len(tags3)
            })
```

```python
merged_df = pd.DataFrame(merged)
df = pd.concat([merged_df, df_l], ignore_index=True)
print(df)
```

```python
memo = {}
for index, row in df.iterrows():
    tag_str = row['Tags'] 
    for tag in tag_str:
        if(tag in memo):
            memo[tag].append(index)
        else:
            memo[tag] = [index]
print(memo)
```

```python
def calculate_frequency(items, n , output_chunk = 100):
    frequency = {}
    for item in items:
        frequency[item] = frequency.get(item, 0) + 1
    sorted_items = sorted(frequency.items(), key=lambda x: (abs(x[1] - n), -x[1]))
    closest_keys = [key for key, _ in sorted_items[:output_chunk]]
    return closest_keys
```

```python
#here res is list, could be used to create frequency list for optimization
def get_candidate_indices(current_index, memo, df):
    
    set_data = df.iloc[current_index]["Tags"]
    res = []
    for tag in set_data:
        rows = memo[tag]
        res += rows
    n = len(set_data) // 2 
    res = calculate_frequency(res, n)
    return res
```

```python
def reorder_df_with_candidates(df, memo):
    df = df.copy().reset_index(drop=True)
    result = []
    unvisited = set(range(len(df)))
    current_index = 0
    unvisited.remove(current_index)
    result.append(df.loc[current_index])
    total_score = 0
    while unvisited:
        curr_tags = df.loc[current_index, 'Tags']
        candidates = set(get_candidate_indices(current_index, memo, df))
        candidates = candidates & unvisited
        if not candidates:
            candidates = unvisited
        candidates = set(itertools.islice(candidates, 500))

        max_score = 0
        next_index = None

        for i in candidates:
            next_tags = df.loc[i, 'Tags']
            common = curr_tags & next_tags
            uncommon1 = (curr_tags - next_tags) 
            uncommon2 = (next_tags - curr_tags)
            score = min(len(common), len(uncommon1), len(uncommon2))

            if score > max_score:
                max_score = score
                next_index = i

        total_score += max_score
        print(current_index, " :::  ", max_score, " ::", total_score)
        if next_index is not None and next_index in unvisited:
            unvisited.remove(next_index)
            result.append(df.loc[next_index])
            current_index = next_index
        else:
            next_index = unvisited.pop()
            result.append(df.loc[next_index])
            current_index = next_index

    return pd.DataFrame(result).reset_index(drop=True)
```

```python
df = reorder_df_with_candidates(df, memo)
print(df)
```

```python
output = df['index']  
len_output = len(output)

with open(output_path, "w") as f:
    f.write(str(len_output) + "\n")
    for line in output.values:
        f.write(str(line)+'\n')
```
